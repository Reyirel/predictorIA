```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2, figsize=(10,8))

# Gráfico de Diabetes
diabetes_counts = df['Diabetes'].value_counts().values
axs[0,0].bar([0,1], diabetes_counts)
axs[0,0].set_title('Diabetes')
axs[0,0].set_xticks([0,1])
axs[0,0].set_xticklabels(['No','Si'])
axs[0,0].text(0, diabetes_counts[0], str(diabetes_counts[0]), color='black', ha='center')
axs[0,0].text(1, diabetes_counts[1], str(diabetes_counts[1]), color='black', ha='center')

# Gráfico de Cáncer
cancer_counts = df['Cáncer'].value_counts().values
axs[0,1].bar([0,1], cancer_counts)
axs[0,1].set_title('Cáncer')
axs[0,1].set_xticks([0,1])
axs[0,1].set_xticklabels(['No','Si'])
axs[0,1].text(0, cancer_counts[0], str(cancer_counts[0]), color='black', ha='center')
axs[0,1].text(1, cancer_counts[1], str(cancer_counts[1]), color='black', ha='center')

# Gráfico de Cirrosis
cirrosis_counts = df['EnfermedadHepática'].value_counts().values
axs[1,0].bar([0,1], cirrosis_counts)
axs[1,0].set_title('Cirrosis')
axs[1,0].set_xticks([0,1])
axs[1,0].set_xticklabels(['No','Si'])
axs[1,0].text(0, cirrosis_counts[0], str(cirrosis_counts[0]), color='black', ha='center')
axs[1,0].text(1, cirrosis_counts[1], str(cirrosis_counts[1]), color='black', ha='center')

# Gráfico de Enfermedad de Tiroides
tiroides_counts = df['EnfermedadTiroidea'].value_counts().values
axs[1,1].bar([0,1], tiroides_counts)
axs[1,1].set_title('Enfermedad Tiroidea')
axs[1,1].set_xticks([0,1])
axs[1,1].set_xticklabels(['No','Si'])
axs[1,1].text(0, tiroides_counts[0], str(tiroides_counts[0]), color='black', ha='center')
axs[1,1].text(1, tiroides_counts[1], str(tiroides_counts[1]), color='black', ha='center')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('limpios.csv')

df = df[['age', 'diab']]
df = df.dropna()

# Gráfico de densidad
fig, ax = plt.subplots()

sns.kdeplot(df.loc[df['diab'] == 0, 'age'], shade=True, label='Sin diabetes', ax=ax)
sns.kdeplot(df.loc[df['diab'] == 1, 'age'], shade=True, label='Con diabetes', ax=ax)

ax.set_xlabel('Edad')
ax.set_ylabel('Densidad')
ax.set_title('Densidad de edad según diabetes')
ax.legend()

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('limpios.csv')

df = df[['age', 'f_cama', 'cama']]
df = df.dropna()

# Gráfico de densidad
fig, ax = plt.subplots()

sns.kdeplot(df.loc[(df['f_cama'] == 0) & (df['cama'] == 0), 'age'],
            fill=True, alpha=0.5,
            label='Sin historial familiar y sin cáncer', ax=ax)

sns.kdeplot(df.loc[(df['f_cama'] == 1) & (df['cama'] == 0), 'age'],
           fill=True, alpha=0.5,
           label='Con historial familiar y sin cáncer', ax=ax)

sns.kdeplot(df.loc[(df['f_cama'] == 0) & (df['cama'] == 1), 'age'],
            fill=True, alpha=0.5,
            label='Sin historial familiar y con cáncer', ax=ax)

sns.kdeplot(df.loc[(df['f_cama'] == 1) & (df['cama'] == 1), 'age'],
            fill=True, alpha=0.5,
            label='Con historial familiar y con cáncer', ax=ax)

ax.set_xlabel('Edad')
ax.set_ylabel('Densidad')
ax.set_title('Densidad de edad según historial de cáncer')
ax.legend()

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('limpios.csv')

df = df[['age', 'diab', 'hypertn']]
df = df.dropna()

# Gráfico de densidad
fig, ax = plt.subplots()

# Sin diabetes
sns.kdeplot(df.loc[(df['diab'] == 0) & (df['hypertn'] == 0), 'age'],
            label='Sin diabetes ni hipertensión', ax=ax,
            fill=True, alpha=0.5, bw=2, color='tab:blue')

sns.kdeplot(df.loc[(df['diab'] == 0) & (df['hypertn'] == 1), 'age'],
            label='Sin diabetes con hipertensión', ax=ax,
            fill=True, alpha=0.5, bw=2, color='tab:orange')

# Con diabetes
sns.kdeplot(df.loc[(df['diab'] == 1) & (df['hypertn'] == 0), 'age'],
            label='Con diabetes sin hipertensión', ax=ax,
            fill=True, alpha=0.5, bw=2, color='tab:green')

sns.kdeplot(df.loc[(df['diab'] == 1) & (df['hypertn'] == 1), 'age'],
            label='Con diabetes e hipertensión', ax=ax,
            fill=True, alpha=0.5, bw=2, color='tab:red')

ax.set_xlabel('Edad')
ax.set_ylabel('Densidad')
ax.set_title('Densidad de edad según diabetes e hipertensión')
ax.legend()

plt.show()
```
