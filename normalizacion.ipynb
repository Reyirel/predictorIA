```python
import pandas as pd

# Cargar el DataFrame
df = pd.read_csv('./imputacion_finals.csv')

# Mostrar las primeras filas del DataFrame para revisar los datos
print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Inicializar los escaladores
minmax_scaler = MinMaxScaler()
standard_scaler = StandardScaler()

# Aplicar la normalización min-max
df_minmax = pd.DataFrame(minmax_scaler.fit_transform(df), columns=df.columns)

# Aplicar la estandarización
df_standard = pd.DataFrame(standard_scaler.fit_transform(df), columns=df.columns)

# Opcional: Guardar los DataFrames normalizados en archivos CSV
df_minmax.to_csv('df_minmax_normalized_overs.csv', index=False)
```
