```python
# Cargar el banco de datos llamado limpios
import pandas as pd

df = pd.read_csv('limpios.csv')
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer, SimpleImputer

# Supongamos que df es tu DataFrame

# Identificar columnas numéricas y categóricas
columnas_numericas = ['weight', 'height', 'waist', 'hip', 'totmet', 'ener_kcal', 'fv', 'dairy', 'ssb', 'rmeat', 'whole_gr', 'multiv', 'parity','afb']
columnas_categoricas = ['maritals', 'indig', 'healths', 'smk', 'oc', 'f_hypertn', 'f_dm', 'f_mi', 'f_cama','pmh','diab','cama','liverd','thyrd']

# Rellenar datos numéricos con KNN
imputer_num = KNNImputer(n_neighbors=5)
df[columnas_numericas] = imputer_num.fit_transform(df[columnas_numericas])

# Rellenar datos categóricos con la moda
imputer_cat = SimpleImputer(strategy='most_frequent')
df[columnas_categoricas] = imputer_cat.fit_transform(df[columnas_categoricas])

# Calcular el BMI si es posible, asumiendo que weight está en kg y height en cm
df.loc[df['bmi'].isnull(), 'bmi'] = df.loc[df['bmi'].isnull(), 'weight'] / ((df.loc[df['bmi'].isnull(), 'height'] / 100) ** 2)

print(df.isnull().sum())  # Verificar los valores nulos después del proceso
```

```python
df.to_csv('imputacion_finals.csv', index=False)
```
